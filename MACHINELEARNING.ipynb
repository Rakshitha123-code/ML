```python
# loading csv files
import pandas as pd
#df=pd.read_csv("filename.csv")
# CREATING A MESSY DATASET
data = {
    "Name": ["A", "B", "C", "D", "E", "E"],
    "Age": [20, 21, None, 22, 20, 20],
    "Hours": [2, 4, 5, None, 8, 8],
    "Marks": [40, 50, 55, 70, None, None]
}

df = pd.DataFrame(data)

print(df)
print(df.isnull())  # to check the null values in the dataset
print(df.isnull().sum())  # to check the number of null values in each column
print(df.isna())  # to check the null values in the dataset
# to remove null values
df=df.dropna()
print(df)
# fill missing values with mean /median
df["Age"]=df["Age"].fillna(df["Age"].mean())
print(df[["Age"]])
print(df.duplicated())  # to find if duplicate values exist
print(df.duplicated().sum())  # to find sum of duplicate values
# remove duplicate values
df=df.drop_duplicates()
print(df.duplicated().sum())
#nsort values
df=df.sort_values("Marks")
print(df)
# sort in descending order
df=df.sort_values("Marks",ascending=False)
print(df)
# reset index
df=df.reset_index(drop=True)
print(df)
# rename columns
df=df.rename(columns={"Hours":"STUDYHOURS"})
print(df.columns)
```
